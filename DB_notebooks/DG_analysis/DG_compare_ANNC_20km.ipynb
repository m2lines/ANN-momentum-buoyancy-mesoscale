```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
```

```python
list = ['0p0','1p0','1p5','2p5','3p5','4p5']

os_dic = {}
ds_mean_dic = {}
#path = '/home/db194/scratch_db194/mom6/base_experiments_2025/DoubleGyre/31Mar/runs/res_20km_ANN_0.0/OUTPUT/'
#os_dic['0p0'] = xr.open_dataset(path + 'ocean.stats.nc', decode_times=False)

for C in list:
    path = f'/home/db194/scratch_db194/mom6/dec2025/DoubleGyre/20km/C{C}/OUTPUT/'
    
    os_dic[C] = xr.open_dataset(path + 'ocean.stats.nc', decode_times=False)
    ds_mean_dic[C] = xr.open_mfdataset(path + 'ave_prog_1*.nc', decode_times=False)
```

```python
plt.figure(figsize=(8,4))
for C in list:
    plt.subplot(1,2,1)
    os_dic[C]['APE'].sum('Interface').plot(label=f'C={C}')
    plt.title('Total APE')
    
    plt.subplot(1,2,2)
    os_dic[C]['KE'].sum('Layer').plot(label=f'C={C}')
    plt.title('Total KE')
    plt.legend()
    
plt.tight_layout()
```
