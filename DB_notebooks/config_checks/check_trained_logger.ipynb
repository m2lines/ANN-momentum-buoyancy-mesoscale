```python
ds = xr.open_dataset(file_path)
```

```python
depth =2 
for factor in ds.factor.values:
    
    ds.MSE_train.sel(factor=factor, depth=depth).plot()
    ds.MSE_validate.sel(factor=factor, depth=depth).plot(label='Validate MSE'+str(factor))
plt.legend()
```
